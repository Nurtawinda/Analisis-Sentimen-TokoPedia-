# Melakukan scrapping ulasan dari aplikasi Tokopedia di Google Play Store

Data ulasan yang diambil dari aplikasi Tokopedia merupakan ulasan yang berasal dari Indonesia dan menggunakan Bahasa Indonesia, dengan jumlah maksimum ulasan yang ingin diambil adalah 10000

```python
scrapview = reviews_all(
    'com.tokopedia.tkpd',
    lang='id',
    country = 'id',
    sort=Sort.MOST_RELEVANT,
    count = 25000
)
```

# Menyimpan ulasan dalam file CSV

Ulasan yang diperoleh dari scraping akan disimpan dalam file CSV bernama **ulasan_tokped.csv**

```python
import csv
with open('ulasan_tokped.csv', mode='w', newline='', encoding='utf-8') as file:
  writer = csv.writer(file)
  writer.writerow(['Review'])
  for review in scrapview:
    writer.writerow([review['content']])
```
