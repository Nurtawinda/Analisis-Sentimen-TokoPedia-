```python
import pickle
import numpy as np
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

# Melakukan Load model yang telah disimpan sebelumnya

```python
model = load_model("model_rnn.h5")

with open("tokenizer.pkl", "rb") as f:
    tokenizer = pickle.load(f)

with open("label_encoder.pkl", "rb") as f:
    le = pickle.load(f)
```

Pada bagian ini, model yang telah disimpan selanjutnya di panggil kembali. Selanjutnya adalah membuat fungsi prediksi yang menggunakan model rnn yang telah di load sebelumnya

```python
max_len = 100

def predict_sentiment(text):
    seq = tokenizer.texts_to_sequences([text])
    padded = pad_sequences(seq, maxlen=max_len, padding='post')

    pred = model.predict(padded)
    label_index = np.argmax(pred)

    label = le.inverse_transform([label_index])[0]

    return label
```

Terakhir adalah melakukan prediksi, dimana kalimat komentar dimasukkan untuk input kemudian model akan melakukan dan memberikan prediksi sesuai dengan hasil pelatihan yang telah dilakukan

```python
text = "Fitur-fiturnya sangat jelek"
print("Teks:", text)
print("Sentimen:", predict_sentiment(text))
```
